```python
# Practice coding here
def kth_character(k:int, word='a') -> str:
    if len(word) >= k:
        return word[k-1]
    else:
        # new_word = [chr(((ord(ch) - 96)%26) + 97) for ch in word]
        new_word = [chr(ord(ch)+1) if ch != 'z' else 'a' for ch in word]
    word += "".join(new_word)
    return kth_character(k, word)

kth_character(10)
```
